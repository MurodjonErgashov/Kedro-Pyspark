# Utils

Small reusable helpers shared across stage notebooks. Run this with `%run ./00_utils.ipynb`, after `00_config.ipynb`.

```python
import pandas as pd


def standardize_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Lowercase/snake_case all column names."""
    df = df.copy()
    df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]
    return df


def clean_text(series: pd.Series) -> pd.Series:
    return series.str.strip()
```
